```python
# définition des noyaux

def rectangular_kernel(u) :
    return 0.5*(np.abs(u) < 1)

def gaussian_kernel(u):
    return((1/np.sqrt(2*np.pi))*np.exp(-u**2/2))
```

```python
def kernel_density1(t,X,h):
    n = len(t)
    Y = np.zeros(n)
    for i in range(n):
        Y[i] = np.mean(gaussian_kernel((t[i] - X)/h)/h)
    return(Y)

def kernel_density2(t,X,h):
    n = len(t)
    Y = np.zeros(n)
    for i in range(n):
        Y[i] = np.mean(rectangular_kernel((t[i] - X)/h)/h)
    return(Y)
    
t = np.linspace(-2,5,101)
plt.plot(t,kernel_density1(t,X,0.5), label='gaussian kernel')
plt.plot(t,kernel_density2(t,X,0.5), label='rectangular kernel')
plt.legend()
plt.show()
```
